### Exercice 1
Créer une class Pile en Python avec ces 5 opérations. Vous utiliserez les listes Python.

```python
class Pile:
    def __init__(self):
        self.pile = []

    def pile_vide(self):
        if self.pile != []:
            return False
        else:
            return True

    def empiler(self,n):
        self.pile.append(n)

    def depiler(self):
        val = self.pile.pop()
        return val

    def sommet(self):
        if len(self.pile) == 0:
            return 0
        else:
            return self.pile[-1]

    def taille(self):
        return len(self.pile)

    def __str__(self) -> str:
        return str(self.pile)

pile = Pile()

pile.empiler(2)
pile.empiler(4)
pile.empiler(5)
pile.empiler(2)

print(pile.depiler())
print(pile)
print(pile.taille())
```

Écrire une procédure ``inverse`` qui prend en paramètre une pile ``p`` et inverse l'ordre de ses éléments (et ne renvoie rien, ce n''est pas une fonction). La pile ``p`` passée en paramètre est modifiée par l'exécution de la procédure.

```python
class Pile:
    def __init__(self):
        self.pile = []

    def pile_vide(self):
        if self.pile != []:
            return False
        else:
            return True

    def empiler(self,n):
        self.pile.append(n)

    def depiler(self):
        val = self.pile.pop()
        return val

    def sommet(self):
        if len(self.pile) == 0:
            return 0
        else:
            return self.pile[-1]

    def taille(self):
        return len(self.pile)

    def __str__(self) -> str:
        return str(self.pile)

class File:
    def __init__(self) -> None:
        self.file = []

    def file_vide(self):
        if self.file != []:
            return False
        else:
            return True

    def enfiler(self,n):
        self.file.append(n)
    
    def defiler(self):
        val = self.file.pop(0)
        return val

    def premier(self):
        return self.file[0]

    def taille(self):
        return len(self.file)

    def __str__(self) -> str:
        return str(self.file)


pile = Pile()

pile.empiler(2)
pile.empiler(4)
pile.empiler(5)
pile.empiler(2)

def inverse(pile):
    lst = []
    for i in range(pile.taille()):
        lst.append(pile.depiler())
    pile.pile = lst
    return pile

def inverse_prof(p1):
    f = File()
    while not p1.pile_vide():
        f.enfiler(p1.depiler())
    while not f.file_vide():
        p1.empiler(f.defiler)
    

print(pile)
print(inverse(pile))
print(inverse_prof(pile))
```

### Exercice 4

### Exercice 5

Dans les deux piles renvoyées, les éléments doivent être dans l'ordre initial, et la pile initial ne doit pas être vide.

```python
class Pile:
    def __init__(self):
        self.pile = []

    def pile_vide(self):
        if self.pile != []:
            return False
        else:
            return True

    def empiler(self,n):
        self.pile.append(n)

    def depiler(self):
        val = self.pile.pop()
        return val

    def sommet(self):
        if len(self.pile) == 0:
            return 0
        else:
            return self.pile[-1]

    def taille(self):
        return len(self.pile)

    def __str__(self) -> str:
        return str(self.pile)

class File:
    def __init__(self) -> None:
        self.file = []

    def file_vide(self):
        if self.file != []:
            return False
        else:
            return True

    def enfiler(self,n):
        self.file.append(n)
    
    def defiler(self):
        val = self.file.pop(0)
        return val

    def premier(self):
        return self.file[0]

    def taille(self):
        return len(self.file)

    def __str__(self) -> str:
        return str(self.file)

def copie(p1):
    p2 = Pile()
    p3 = Pile()
    while not p1.pile_vide():
        p2.empiler(p1.depiler())
    while not p2.pile_vide():
        elt = p2.depiler()
        p1.empiler(elt)
        p3.empiler(elt)
    return p3

def pair_impair(p):
    p_pair = Pile()
    p_impair = Pile()
    p_inter = Pile()
    while not p.pile_vide():
        p_inter.empiler(p.depiler())
    while not p_inter.pile_vide():
        elt = p_inter.depiler()
        if elt % 2 == 0 :
            p_pair.empiler(elt)
        else:
            p_impair.empiler(elt)
    return p_pair, p_impair

p = Pile()
for i in [2,4,5,2,1,5,6,8]:
    p.empiler(i)

p_pair, p_impair = pair_impair(p)
print(p_pair)
print(p_impair)
```

### Exercice 6
On dispose d'une chaîne de caractère contenant uniquement des parenthèses ouvrantes et fermantes, comme "((())))". Un parenthésage est correct, si en parcourant la chaîne de caractère de gauche à droite, à tout moment , le nombre de parenthèses dejà ouvertes est supérieur ou égal au nombre de parenthèses déjà fermées. De plus, dans la chaîne complète, il doit y avoir autant de parenthèses ouvrantes que de parenthèses fermantes.
1. Écrire une portion de code, qui en utilisant le type Pile, parcourt la chaîne de caractère et empile une valeur (peu importe laquelle) lorsque le caractère est "(" et dépile une valeur lorsque le caractère est ")".
2. Tester le code précédent sur les chaînes : "(()())", "())()(" et "(()". Que dire de ces trois cas ? Et comment, avec la pile, détecter le bon parenthésage ?
3. Écrire une fonction ``parenthesage`` qui prend en paramètre une chaîne de caractères et renvoie ``True`` si la chaîne est bien parenthésée et ``False`` sinon. Attention, la fonction devra renvoyer ``False`` si la chaîne de caractère comporte un autre caractère que '(' et ')'.

```python
class Pile:
    def __init__(self):
        self.pile = []

    def pile_vide(self):
        if self.pile != []:
            return False
        else:
            return True

    def empiler(self,n):
        self.pile.append(n)

    def depiler(self):
        val = self.pile.pop()
        return val

    def sommet(self):
        if len(self.pile) == 0:
            return 0
        else:
            return self.pile[-1]

    def taille(self):
        return len(self.pile)

    def __str__(self) -> str:
        return str(self.pile)

chaine = "(5+2) * 3"

def parenthesage(chaine):
    p = Pile()
    for carac in chaine:
        if carac == '(':
            p.empiler(carac)
        elif carac == ')':
            if not p.pile_vide():
                p.depiler()
            else:
                return False
    return p.pile_vide()

print(parenthesage(chaine))
```

```python
class Pile:
    def __init__(self):
        self.pile = []

    def pile_vide(self):
        if self.pile != []:
            return False
        else:
            return True

    def empiler(self,n):
        self.pile.append(n)

    def depiler(self):
        val = self.pile.pop()
        return val

    def sommet(self):
        if len(self.pile) == 0:
            return 0
        else:
            return self.pile[-1]

    def taille(self):
        return len(self.pile)

    def __str__(self) -> str:
        return str(self.pile)

class File:
    def __init__(self) -> None:
        self.file = []

    def file_vide(self):
        if self.file != []:
            return False
        else:
            return True

    def enfiler(self,n):
        self.file.append(n)
    
    def defiler(self):
        val = self.file.pop(0)
        return val

    def premier(self):
        return self.file[0]

    def taille(self):
        return len(self.file)

    def __str__(self) -> str:
        return str(self.file)

class Maillon:
    def __init__(self,valeur=None):
        self.val = valeur
        self.suiv = None # Pas de maillon suivant

    def __str__(self):
        return f"({self.val})" # affiche simplement la valeur du maillon

class ListeChainee:
    def __init__(self):
        self.tete = None # Liste vide

    def __str__(self) -> str:
        affichage = []
        maillon = self.tete
        while maillon is not None:
            affichage.append(str(maillon.val))
            maillon = maillon.suiv
        return "{" + " ; ".join(affichage) + "}"   

    def est_vide(self):
        return self.tete == None
    
    def taille(self):
        maillon = self.tete
        l = 0
        while maillon is not None:
            maillon = maillon.suiv
            l += 1
        return l

    def get_dernier_maillon(self) -> Maillon | None:
        maillon = self.tete
        last_maillon = None
        while maillon is not None:
            last_maillon = maillon
            maillon = maillon.suiv
        return last_maillon

    def get_maillon_indice(self,i) -> Maillon | None:
        maillon = self.tete
        if i < 0:
            return None
        for j in range(i):
            maillon = maillon.suiv #type: ignore
        return maillon

    def ajouter_maillon_fin(self,valeur):
        maillon_fin = Maillon(valeur)
        if self.est_vide():
            self.tete = maillon_fin
        else:
            dernier_maillon = self.get_dernier_maillon()
            dernier_maillon.suiv = maillon_fin #type: ignore

    def ajouter_maillon_debut(self,valeur):
        nv_tete = Maillon(valeur)
        tete = self.tete
        nv_tete.suiv = tete #type: ignore
        liste.tete = nv_tete

    def ajouter_maillon(self,indice,valeur):
        if indice < 0 or indice > self.taille():
            return None
        elif indice == 0:
            self.ajouter_maillon_debut(valeur)
        else:
            nx_maillon = Maillon(valeur)
            maillon = self.get_maillon_indice(indice-1)
            maillon_apres = maillon.suiv # type: ignore
            maillon.suiv = nx_maillon # type: ignore
            nx_maillon.suiv = maillon_apres

    def supprimer_dernier_maillon(self):
        if self.est_vide():
            return None
        elif self.taille() == 1:
            maillon_renvoie = self.tete
            self.taille = None
            return maillon_renvoie
        else:
            dernier_maillon = self.get_dernier_maillon()
            avant_dernier_maillon = self.get_maillon_indice(self.taille()-2)
            avant_dernier_maillon.suiv = None # type: ignore
            return dernier_maillon

    def supprimer_premier_maillon(self):
        if self.est_vide():
            return None
        elif self.taille() == 1:
            maillon = self.tete
            self.tete = None
            return maillon
        else:
            premier_maillon = self.tete
            deuxieme_maillon = self.tete.suiv # type: ignore
            self.tete = deuxieme_maillon
            premier_maillon.suiv = None #type: ignore
            return premier_maillon

    def supprimer_maillon(self,indice):
        if indice > self.taille() or indice < 0:
            return None
        elif indice == 0:
            maillon_sup = self.supprimer_premier_maillon()
        elif indice == self.taille()-1:
            maillon_sup1 = self.supprimer_dernier_maillon()
            return maillon_sup1
        else:
            maillon_indice = self.get_maillon_indice(indice)
            maillon_indice_avant = self.get_maillon_indice(indice-1)
            maillon_indice_apres = self.get_maillon_indice(indice+1)
            maillon_indice_avant.suiv = maillon_indice_apres #type: ignore
            maillon_indice.suiv = None #type: ignore
            return maillon_indice

class Pile_Maillon:
    def __init__(self):
        self.pile = ListeChainee()

    def pile_vide(self):
        return self.pile.est_vide()

    def empiler(self,n):
        return self.pile.ajouter_maillon_fin(n)

    def depiler(self):
        val = self.pile.supprimer_dernier_maillon()
        return val

    def sommet(self):
        return self.pile.get_dernier_maillon

    def taille(self):
        return self.pile.taille()

    def __str__(self) -> str:
        return str(self.pile)

class File_Maillon:
    def __init__(self) -> None:
            self.file = ListeChainee()
    
    def file_vide(self):
        return self.file.est_vide()

    def enfiler(self,n):
        self.file.ajouter_maillon_fin(n)
    
    def defiler(self):
        val = self.file.supprimer_premier_maillon()
        return val

    def premier(self):
        return self.file.get_maillon_indice(0)

    def taille(self):
        return self.file.taille()

    def __str__(self) -> str:
        return str(self.file)
```
