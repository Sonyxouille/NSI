### Exercice 1
Créer une class Pile en Python avec ces 5 opérations. Vous utiliserez les listes Python.

```python
class Pile:
    def __init__(self):
        self.pile = []

    def pile_vide(self):
        if self.pile != []:
            return False
        else:
            return True

    def empiler(self,n):
        self.pile.append(n)

    def depiler(self):
        val = self.pile.pop()
        return val

    def sommet(self):
        if len(self.pile) == 0:
            return 0
        else:
            return self.pile[-1]

    def taille(self):
        return len(self.pile)

    def __str__(self) -> str:
        return str(self.pile)

pile = Pile()

pile.empiler(2)
pile.empiler(4)
pile.empiler(5)
pile.empiler(2)

print(pile.depiler())
print(pile)
print(pile.taille())
```

### Exercice 2
Implémentez en Python ces 5 opérations à l'aide de liste en Python.

```python
class File:
    def __init__(self) -> None:
        self.file = []

    def file_vide(self):
        if self.file != []:
            return False
        else:
            return True

    def enfiler(self,n):
        self.file.append(n)
    
    def defiler(self):
        val = self.file.pop(0)
        return val

    def premier(self):
        return self.file[0]

    def taille(self):
        return len(self.file)

    def __str__(self) -> str:
        return str(self.file)

f = File()
print(f.file_vide())
for i in [2,4,5,2]:
    f.enfiler(i)

print(f)
print(f.file_vide())
print(f.taille())
print(f.premier())
print(f.defiler())
print(f)
```
