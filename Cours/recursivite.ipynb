## Exo 1 : Écrire une fonction ```somme(n)``` qui retourne la somme des $n$ premiers entiers.



````python
def somme(n:int)->int:
    """Calcul la somme des n premiers entiers"""
    somme = 0
    for i in range(n+1):
        somme += i
    return somme

somme(3)
````

## Exo 2 : Donner une définition récursive qui correspond au calcul de la fonction factorielle ``n!`` définie par ``n! = 1×2×3×...×n`` si ``n > 0`` et ``0! = 1``. Écrire également le code d'une fonction ``facto`` qui implémente cette définition. Écrire également la version itérative.

$$facto(n) = \begin{cases} 1 & si \quad n = 0 \\ n * facto(n-1) & si \quad n > 0 \end{cases}$$

- $4! = 4*3*2*1$
- $5! = 5*4*3*2*1 = 5 * 4!$

````python
def facto_recursive(n: int) -> int:
    """Calcul la factorielle de manière récursive"""
    if n == 0:
        return 1
    else:
        return n * facto_recursive(n-1)

facto_recursive(5)

def facto_iterative(n:int)->int:
    """Calcul la factorielle de manière itérative"""
    facto = 1
    for i in range(1,n+1):
        facto *= i
    return facto

facto_iterative(5)
````

Fibonacci

````python
def fibonacci_iterative_liste(n:int):
    """calcul la suite de Fibonacci de manière iterative"""
    if n == 0:
        return 0
    elif n == 1:
        return 1
    fibonacci = [0,1]
    for i in range(1,n+1):
        fibonacci.append(fibonacci[i-1]+fibonacci[i])
    return fibonacci

print(fibonacci_iterative_liste(60))

def fibo_ite_nolst(n):
    if n == 0 or n == 1:
        return n
    F0 = 0
    F1 = 1
    for i in range(2,n+1):
        f = F0 + F1
        F0 = F1
        F1 = f
    return f

print(fibo_ite_nolst(5))


def fibonacci_rec(n):
    if n == 0 or n == 1:
        return n
    else: 
        return fibonacci_rec(n-1) + fibonacci_rec(n-2)

print(fibonacci_rec(25))
````

## Exo 3 : Écrire une fonction récursive ``boucle(i,k)`` qui affiche les entiers entre ``i`` et ``k``. Par exemple, ``boucle(0,3)`` affiche 0,1,2,3. Pour les plus rapides, écrire la version itérative.

````python
def boucle_iterative(i,k):
    """Boucle chiffre entre i et k en itératif"""
    les_chiffres = []
    for chiffre in range(i,k+1):
        les_chiffres.append(chiffre)
    return les_chiffres

print(boucle_iterative(0,9))

def boucle_recursive(i,k):
    """Boucle chiffre entre i et k en récursif"""
    if i == k:
        return i
    else:
        print(i)
        return boucle_recursive(i+1,k)

print(boucle_recursive(0,9))
````

## Exo 4 : La méthode du paysan russe est un très vieil algorithme de multiplication de deux nombres entiers déjà décrit, sous forme légèrement différente, sur un papyrus égyptien rédigé autour de 1650 avant J.C. Il s'agissait de la principale méthode de calcul en Europe avant l'introduction des chiffres arabes. Les ordinateurs l'ont utilisé avant que la multiplication ne soit directement intégrée dans le processeur. Sous une forme moderne, il peut être décrit par le programme donné dans l'algorithme ci-dessous.

``` 
(Pseudo-code de la fonction itérative)

Fonction multiplication(x,y):
    p = 0
    TANT QUE x > 0:
        Si x est impair:
            p = p + y
        x = x // 2
        y = y + y
    Retourner p
```

1. Appliquer cette fonction pour effectuer la multiplication de 105 par 253. Détailler les étapes dans le tableau suivant :

|  x  |  y  |  p  | ligne |
|:---:|:---:|:---:|:-----:|

2. On admet que cet algorithme repose sur les relations suivantes :

$$x * y = \begin{cases} 0 & \text{si } n = 0 \\ (x/\!/2) * (y+y) & \text{si } x \text{ est pair} \\ (x/\!/2) * (y+y) + y & \text{si } x \text{ est impair} \end{cases}$$

Proposer une version récursive de cet algorithme.

### 1. 

````python
def fonction_multiplicative(x,y):
    p = 0
    while x > 0:
        if x % 2 == 1:
            p += y
        x = x // 2
        y += y
        print(f"x: {x}, y: {y}, p: {p}")
    return p

fonction_multiplicative(105,253)
````
