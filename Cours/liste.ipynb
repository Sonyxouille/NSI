De même, écrire les méthodes ``get_dernier_maillon()`` qui retourne le dernier maillon de la chaîne, et la méthode ``get_maillon_indice(i)`` qui retourne la maillon d'indice ``i``.

````python
class Maillon:
    def __init__(self,valeur=None):
        self.val = valeur
        self.suiv = None # Pas de maillon suivant

    def __str__(self):
        return f"({self.val})" # affiche simplement la valeur du maillon

class ListeChainee:
    def __init__(self):
        self.tete = None # Liste vide

    def __str__(self) -> str:
        affichage = []
        maillon = self.tete
        while maillon is not None:
            affichage.append(str(maillon.val))
            maillon = maillon.suiv
        return "{" + " ; ".join(affichage) + "}"   

    def est_vide(self):
        return self.tete == None
    
    def taille(self):
        maillon = self.tete
        l = 0
        while maillon is not None:
            maillon = maillon.suiv
            l += 1
        return l

    def get_dernier_maillon(self) -> Maillon | None:
        maillon = self.tete
        last_maillon = None
        while maillon is not None:
            last_maillon = maillon
            maillon = maillon.suiv
        return last_maillon

    def get_maillon_indice(self,i) -> Maillon | None:
        maillon = self.tete
        if i < 0:
            return None
        for j in range(i):
            maillon = maillon.suiv #type: ignore
        return maillon

    def ajouter_maillon_fin(self,valeur):
        maillon_fin = Maillon(valeur)
        if self.est_vide():
            self.tete = maillon_fin
        else:
            dernier_maillon = self.get_dernier_maillon()
            dernier_maillon.suiv = maillon_fin #type: ignore

    def ajouter_maillon_debut(self,valeur):
        nv_tete = Maillon(valeur)
        tete = self.tete
        nv_tete.suiv = tete #type: ignore
        liste.tete = nv_tete

    def ajouter_maillon_dans_ton_crane(self,valeur,indice1):
        maillon_indice1 = self.get_maillon_indice(indice1)
        maillon_indice2 = self.get_maillon_indice(indice1 + 1)
        maillon_ajoute = Maillon(valeur)
        maillon_indice1.suiv = maillon_ajoute #type: ignore
        maillon_ajoute.suiv = maillon_indice2 #type: ignore

liste = ListeChainee()
M1, M2, M3 = Maillon(5), Maillon(8), Maillon(9)

M1.suiv = M2 #type: ignore
M2.suiv = M3 #type: ignore

liste.tete = M1 #type: ignore

print(liste)
print(liste.taille())
print(liste.get_dernier_maillon())
print(liste.get_maillon_indice(3))
print(liste.ajouter_maillon_fin(10))
````
