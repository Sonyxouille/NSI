Sur le modèle du code précédent, écrire la méthode ```taille(self)``` qui retourne la taille de la liste chaînée.

````python
class Maillon:
    def __init__(self,valeur=None):
        self.val = valeur
        self.suiv = None # Pas de maillon suivant

    def __str__(self):
        return f"({self.val})" # affiche simplement la valeur du maillon

class ListeChainee:
    def __init__(self):
        self.tete = None # Liste vide

    def __str__(self) -> str:
        affichage = []
        maillon = self.tete
        while maillon is not None:
            affichage.append(str(maillon.val))
            maillon = maillon.suiv
        return "{" + " ; ".join(affichage) + "}"   
     
    def est_vide(self):
        return self.tete == None
    
    def taille(self):
        maillon = self.tete
        l = 0
        while maillon is not None:
            maillon = maillon.suiv
            l += 1
        return l

    def get_dernier_maillon(self):
        maillon = self.tete
        last_maillon = None
        while maillon is not None:
            last_maillon = maillon
            maillon = maillon.suiv
        return last_maillon

    def get_maillon_indice(self,i):
        maillon = self.tete
        for j in range(i):
            maillon = maillon.suiv
        return maillon   

liste = ListeChainee()
M1, M2, M3 = Maillon(5), Maillon(8), Maillon(9)

M1.suiv = M2 #type: ignore
M2.suiv = M3 #type: ignore

liste.tete = M1 #type: ignore

print(liste)
print(liste.taille())
print(liste.get_dernier_maillon())
print(liste.get_maillon_indice(3))
````
