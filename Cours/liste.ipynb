````python
class Maillon:
    def __init__(self,valeur=None):
        self.val = valeur
        self.suiv = None # Pas de maillon suivant

    def __str__(self):
        return f"({self.val})" # affiche simplement la valeur du maillon

class ListeChainee:
    def __init__(self):
        self.tete = None # Liste vide

    def __str__(self) -> str:
        affichage = []
        maillon = self.tete
        while maillon is not None:
            affichage.append(str(maillon.val))
            maillon = maillon.suiv
        return "{" + " ; ".join(affichage) + "}"   

    def est_vide(self):
        return self.tete == None
    
    def taille(self):
        maillon = self.tete
        l = 0
        while maillon is not None:
            maillon = maillon.suiv
            l += 1
        return l

    def get_dernier_maillon(self) -> Maillon | None:
        maillon = self.tete
        last_maillon = None
        while maillon is not None:
            last_maillon = maillon
            maillon = maillon.suiv
        return last_maillon

    def get_maillon_indice(self,i) -> Maillon | None:
        maillon = self.tete
        if i < 0:
            return None
        for j in range(i):
            maillon = maillon.suiv #type: ignore
        return maillon

    def ajouter_maillon_fin(self,valeur):
        maillon_fin = Maillon(valeur)
        if self.est_vide():
            self.tete = maillon_fin
        else:
            dernier_maillon = self.get_dernier_maillon()
            dernier_maillon.suiv = maillon_fin #type: ignore

    def ajouter_maillon_debut(self,valeur):
        nv_tete = Maillon(valeur)
        tete = self.tete
        nv_tete.suiv = tete #type: ignore
        liste.tete = nv_tete

    def ajouter_maillon(self,indice,valeur):
        if indice < 0 or indice > self.taille():
            return None
        elif indice == 0:
            self.ajouter_maillon_debut(valeur)
        else:
            nx_maillon = Maillon(valeur)
            maillon = self.get_maillon_indice(indice-1)
            maillon_apres = maillon.suiv # type: ignore
            maillon.suiv = nx_maillon # type: ignore
            nx_maillon.suiv = maillon_apres

    def supprimer_dernier_maillon(self):
        if self.est_vide():
            return None
        elif self.taille() == 1:
            maillon_renvoie = self.tete
            self.taille = None
            return maillon_renvoie
        else:
            dernier_maillon = self.get_dernier_maillon()
            avant_dernier_maillon = self.get_maillon_indice(self.taille()-2)
            avant_dernier_maillon.suiv = None # type: ignore
            return dernier_maillon

    def supprimer_premier_maillon(self):
        if self.est_vide():
            return None
        elif self.taille() == 1:
            maillon = self.tete
            self.tete = None
            return maillon
        else:
            premier_maillon = self.tete
            deuxieme_maillon = self.tete.suiv # type: ignore
            self.tete = deuxieme_maillon
            premier_maillon.suiv = None #type: ignore
            return premier_maillon

    def supprimer_maillon(self,indice):
        if indice > self.taille() or indice < 0:
            return None
        elif indice == 0:
            maillon_sup = self.supprimer_premier_maillon()
        elif indice == self.taille()-1:
            maillon_sup1 = self.supprimer_dernier_maillon()
            return maillon_sup1
        else:
            maillon_indice = self.get_maillon_indice(indice)
            maillon_indice_avant = self.get_maillon_indice(indice-1)
            maillon_indice_apres = self.get_maillon_indice(indice+1)
            maillon_indice_avant.suiv = maillon_indice_apres #type: ignore
            maillon_indice.suiv = None #type: ignore
            return maillon_indice


liste = ListeChainee()
M1, M2, M3 = Maillon(5), Maillon(8), Maillon(9)

M1.suiv = M2 #type: ignore
M2.suiv = M3 #type: ignore

liste.tete = M1 #type: ignore

print(liste)
print(liste.taille())
print(liste.get_dernier_maillon())
liste.get_maillon_indice(3)
liste.ajouter_maillon_fin(10)
liste.ajouter_maillon(2,10)
print(liste)
liste.supprimer_dernier_maillon()
print(liste)
````
