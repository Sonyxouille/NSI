## Exo 3 : inversion de dictionnaire
Ecrire une fonction inverser(d) qui inverse les clés et les valeurs d'un dictionnaire

```inverser({"a": 1, "b": 2, "c": 3})```

````python
def inverser(dico:dict)->dict:
    dico2 = {}
    for k,v in dico.items():
        dico2[v] = k
    return dico2

d = {"a": 1, "b": 2, "c": 3}
print(inverser(d))
````

## Exo 5 : dictionnaire par compréhension imbriqué
A partir d'une liste de mots, contruire un dictionnaire où chaque mot est associé à un sous-dictionnaire comptant ses voyelles et consonnes.
```mots = ["python", "dictionnaire", "algorithme"]```

````python
mots = ["python", "dictionnaire", "algorithme"]
voyelle = ["a","e","i","o","u","y"]
dico = {}

for mot in mots:
    dico[mot] = {"voyelles" : 0, "consonnes" : 0}
    for i in mot:
        if i in voyelle:
            dico[mot]["voyelles"] += 1
        else:
            dico[mot]["consonnes"] += 1

print(dico)
````

## Exo 6 : fusion intelligente
Écris une fonction fusionner(d1, d2) qui fusionne deux dictionnaires. Si une clé existe dans les
deux, les valeurs (numériques) sont additionnées.
```
d1 = {"a" : 5, "b" : 3, "c" : 8}
d2 = {"b" : 2, "c" : 1, "d" : 4}

````python
def fusionner(d1:dict,d2:dict)->dict:
    """Fonction qui permet de fusionner 2 dictionnaires"""
    d3 = {}
    for k1,v1 in d1.items():
        for k2,v2 in d2.items():
            if k1 == k2:
                d3[k1] = v1 + v2
            if k1 not in d3:
                d3[k1] = v1
            if k2 not in d3:
                d3[k2] = v2
    return d3

d1 = {"a" : 5, "b" : 3, "c" : 8}
d2 = {"b" : 2, "c" : 1, "d" : 4}

print(fusionner(d1,d2))

def fusionner_prof(d1:dict,d2:dict)->dict:
    d3 = d1.copy()
    for k2,v2 in d2.items():
        if k2 not in d3:
            d3[k2] = v2
        else:
            d3[k2] += v2
    return d3

print(fusionner_prof(d1,d2))
````

## Exo 7 : groupement par critère
Écris une fonction qui groupe une liste de mots par leur première lettre.
```mots = ["ananas", "banane", "abricot", "cerise", "avocat", "brugnon"]```

````python
mots = ["ananas", "banane", "abricot", "cerise", "avocat", "brugnon"]

def liste_premiere_lettre2(lst):
    lst_tries = {}
    for mot in lst:
        if mot[0] in lst_tries:
            lst_tries[mot[0]].append(mot)
        else:
            lst_tries[mot[0]] = [mot]
    return lst_tries

print(liste_premiere_lettre2(mots))
````
