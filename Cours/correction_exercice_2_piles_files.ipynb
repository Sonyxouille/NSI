```python
class Pile:
    def __init__(self):
        self.pile = []

    def pile_vide(self):
        if self.pile != []:
            return False
        else:
            return True

    def empiler(self, n):
        self.pile.append(n)

    def depiler(self):
        val = self.pile.pop()
        return val

    def sommet(self):
        if len(self.pile) == 0:
            return 0
        else:
            return self.pile[-1]

    def taille(self):
        return len(self.pile)

    def __str__(self) -> str:
        return str(self.pile)
```

#### 4.a.

```python
def est_vide(f):
    return f[0].pile_vide() and f[1].pile_vide()
```

#### 4.b.

```python
def enfiler(f, elt):
    f[0].empiler(elt)
```

#### 4.c.

```python
def defiler(f):
    if est_vide(f):
        return None

    if f[1].pile_vide():
        while not f[0].pile_vide():
            elt = f[0].depiler()
            f[1].empiler(elt)

    return f[1].depiler()
```
