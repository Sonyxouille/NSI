### 6) Réaliser une fonction qui détermine le nombre d'apparition d'une lettre dans une chaine de caractère. Vous réaliserez les tests nécessaires.

````python
def nombre_lettre(phrase):
    dico = {}
    for i in phrase:
        if i in dico:
            dico[i] += 1
        else:
            dico[i] = 1
    return dico

def nombre_lettre_correction(chaine2:str,lettre2:str)->int:
    """ Fonction qui calcule le nombre d'apparition d'une lettre dans une chaine"""
    assert type(chaine2) == str,"chaine2 n'est pas une chaine de caractère"
    assert type(lettre2) == str, "lettre2 n'est pas une lettre"
    nb = 0
    for i in range(len(chaine2)):
        if lettre2 == chaine2[i]:
            nb += 1
    return nb

chaine = "chaine de caractère"
lettre = "e"

print(nombre_lettre(chaine))
print(nombre_lettre_correction(chaine,lettre))
````

### 7) Réaliser une fonction qui réalise la moyenne des éléments d'une liste.

````python
def moyenne_liste(lst:list)-> float:
    """Fonction qui fait la moyenne des chiffre d'une liste"""
    moyenne = 0
    for i in lst:
        moyenne += i
    return moyenne / len(lst)

liste_nombre = [1,2,3,4,5]

print(moyenne_liste(liste_nombre))
````
