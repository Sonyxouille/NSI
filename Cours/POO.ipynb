## Cours

### Programme 1

```python
class Chrono:
    "une classe pour représenter le temps mesuré en heures, minutes et secondes"

    def __init__(self, h, m, s):
        "le constructeur prend comme arguments temps, minutes et secondes"
        self.heures = h
        self.minutes = m
        self.secondes = s

    def __str__(self):
        """ affiche les heures, minutes, secondes"""
        return (str(self.heures) + "h"
                + str(self.minutes) + "m"
                + str(self.secondes) + "s")
    
t = Chrono(9,12,30)
print(t)

s = Chrono(19,48,60)
print(s)
print(s.__doc__)
print(s.__init__.__doc__)
```

### Exercice 1

Écrire une classe Dinosaure dont les attributs sont :

- sa longueur ;
- sa hauteur ;
- son poids ;
- sa vitesse maximale

Vous construirez l’instance triceratops avec comme caractéristiques : 9 m de longueur, 3 m de hauteur, une masse de 9 tonnes environ, et une vitesse maximale donnée à 32 km/h. Puis l’instance Tyrannosaurus Rex. Celui-ci mesurait 13 m de longueur, pesait 8 t environ, 6 m de haut et devait courir à 27 km/h.

```python
class Dinosaure:
    "Une classe permettant de présenter la longueur, la hauteur, le poids et la vitesse maximale d'un dinosaure"

    def __init__(self,l,h,pds,kmh):
        "Le constructeur prend comme argument longueur, hauteur, poids et vitesse maximale"
        self.longueur = l
        self.hauteur = h
        self.poids = pds
        self.vitesse_max = kmh

    def __str__(self):
        "Affiche ses caractéristiques"
        return (str(self.longueur) + "m de longueur, "
                + str(self.hauteur) + "m de hauteur, "
                + "une masse de " + str(self.poids) + "tonnes environ"
                + "et une vitesse maximale donnée à " + str(self.vitesse_max) + "km/h.")

triceratops = Dinosaure(9,3,9,32)
print(triceratops)

tyrannosaurus_rex = Dinosaure(13,6,8,27)
print(tyrannosaurus_rex)
print(tyrannosaurus_rex.__doc__)
print(tyrannosaurus_rex.__init__.__doc__)
```

```python
class DinosaureProf:
    "Une classe permettant de présenter la longueur, la hauteur, le poids et la vitesse maximale d'un dinosaure"

    def __init__(self,l,h,pds,kmh):
        "Le constructeur prend comme argument longueur, hauteur, poids et vitesse maximale"
        self.longueur = l
        self.hauteur = h
        self.poids = pds
        self.vitesse_max = kmh

    def __str__(self):
        "Affiche ses caractéristiques"
        return f"Ce dinosaure a {self.hauteur} m de hauteur"

triceratops = DinosaureProf(9,3,9,32)
print(triceratops)

tyrannosaurus_rex = Dinosaure(13,6,8,27)
print(tyrannosaurus_rex)
print(tyrannosaurus_rex.__doc__)
print(tyrannosaurus_rex.__init__.__doc__)
```
